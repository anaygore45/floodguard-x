# Geospatial Intelligence and Terrain Analysis

## Objective
Extract OSM drainage network, examine Copernicus DEM topography, and map spatial distribution of chronic waterlogging spots across municipal administrative wards.

## Input Datasets
- data/raw/osm_mumbai_central_corridor.osm
- data/raw/copernicus_dem_mumbai_hotspots.csv
- data/raw/mcgm_chronic_waterlogging_hotspots.csv

---

### Step 1: Load Hotspots and DEM Elevation Profiles

```python
import os
import pandas as pd

base_dir = os.path.abspath(os.path.join(os.getcwd(), "..")) if "notebooks" in os.getcwd() else os.getcwd()
hotspots_path = os.path.join(base_dir, "data", "raw", "mcgm_chronic_waterlogging_hotspots.csv")
df_hotspots = pd.read_csv(hotspots_path)

print(f"Loaded {len(df_hotspots)} verified MCGM chronic waterlogging spots.")
print(df_hotspots[["hotspot_id", "name", "ward", "elevation_m", "risk_tier", "catchment"]].head(10))
```

### Step 2: Elevation Profile and Ward Distribution

```python
print("Topographic Elevation Distribution (meters above MSL):")
print(df_hotspots["elevation_m"].describe())

ward_counts = df_hotspots["ward"].value_counts()
print("Hotspots by MCGM Administrative Ward:")
print(ward_counts)
```

### Step 3: Inspect OpenStreetMap Extract Header & Features

```python
osm_path = os.path.join(base_dir, "data", "raw", "osm_mumbai_central_corridor.osm")
with open(osm_path, "r", encoding="utf-8") as f:
    sample_lines = [f.readline() for _ in range(15)]
print("OSM XML Sample Header:")
print("".join(sample_lines[:10]))
```

---
## Analytical Limitations & Assumptions
Public OSM does not map subterranean storm sewers; analysis uses surface drainage, major nullahs, and Copernicus 30m/90m topography.

## Key Conclusions
Most critical flood spots cluster in low-lying depressions with elevation < 7m MSL and close proximity to tidal nullahs (Mithi, Poisar, Dahisar).
